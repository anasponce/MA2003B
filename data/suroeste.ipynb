```python
import pandas as pd
from concat_helper import concat_by_station
station_name = "SUROESTE"   
file_2022_2023 = "DATOS HISTÓRICOS 2022_2023_TODAS ESTACIONES.xlsx"
file_2023_2024 = "DATOS HISTÓRICOS 2023_2024_TODAS ESTACIONES_ITESM.xlsx"

df_final = concat_by_station(file_2022_2023, file_2023_2024, station_name)


assert not (df_final["date"].dt.year == 1970).any(), "Quedó alguna fecha mal parseada."
assert df_final.columns.is_unique, "Hay columnas duplicadas."

print(df_final.head())
print(df_final.tail())
print(df_final["date"].min(), "→", df_final["date"].max())
```

```python
rango_completo = pd.date_range(start=df_final['date'].min(),
                               end=df_final['date'].max(),
                               freq='H')

horas_faltantes = rango_completo.difference(df_final['date'])

print("Cantidad de horas faltantes:", len(horas_faltantes))
print("Ejemplos de horas faltantes:\n", horas_faltantes[:10])
```
